###  عملگر های ریاضی
خب برای ضرب و تقسیم و جمع و تفریق مشخصا توی پایتون یکسری عملگر وجود داره

از + برای جمع و
از - برای تفریق و
از * برای ضرب و
از / برای تقسیم و
از ** برای به توان رسوندن استفاده میکنیم.

خب اینها عملگر های معمولی هستن که همه آشنها هستیم و همه روزه استفادشون میکنیم.

```python
math_operation = 10 / 4
another_math_operation = 23 * 14
print(math_operation)
print(another_math_operation)
```

 ### عملگر های که یکم عجیب ترن

یکیشون % هستش که برای باقی مانده یک تقسیمه

و دیگری // که تقسیم صحیح هستش یعنی میاد تقسیم میکنه و سپس فقط بخش صحیح رو برمیگردونه

```python
even = 23 % 2
print(even)
my_new_operation = 35 // 2
print(my_new_operation)
```
